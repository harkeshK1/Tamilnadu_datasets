```python
def greet_user(name, times):
  """Prints a greeting to the user a specified number of times."""
  print(f"--- Greeting {times} times for {name} ---")
  for i in range(times):
    print(f"Hello, {name}! (Time {i + 1})")

# Call the function with sample arguments
greet_user("Alex", 3)
```

```python
import pandas as pd
import os

# Folder path
folder_path = r"E:\healthcare\datasets\socio economic data"

# Get all CSV files
csv_files = [f for f in os.listdir(folder_path) if f.endswith(".csv")]

# Read and combine
dataframes = []
for file in csv_files:
    file_path = os.path.join(folder_path, file)
    try:
        df = pd.read_csv(file_path)
        df["SourceFile"] = file  # Optional: track which state file each row came from
        dataframes.append(df)
        print(f"Loaded: {file}")
    except Exception as e:
        print(f"❌ Error loading {file}: {e}")

# Combine all into one DataFrame
if dataframes:
    combined_df = pd.concat(dataframes, ignore_index=True)
    print(f"\n✅ Combined {len(dataframes)} files successfully!")
    print("Shape:", combined_df.shape)

    # Save the merged dataset
    output_path = os.path.join(folder_path, "combined_gdp_data.csv")
    combined_df.to_csv(output_path, index=False)
    print(f"📁 Saved combined file at: {output_path}")
else:
    print("⚠️ No CSV files found or could not be combined.")

```

```python
import pandas as pd

# === FILE PATHS ===
files = {
    "climate": r"E:\healthcare\datasets\climate vulnerability\7160_KEYS.csv",
    "health": r"E:\healthcare\datasets\family health survey\national_family_health_survey.csv",
    "household_gdp": r"E:\healthcare\datasets\india gdp houshold\GDP and Household Consumption Expenditure Data for India (1990-2019).xlsx",
    "hospitals": r"E:\healthcare\datasets\india hospitals\Hospitals In India (Anonymized).csv",
    "expenses": r"E:\healthcare\datasets\monthly expense\Monthly_Expense_Data.csv",
    "state_gdp": r"E:\healthcare\datasets\state gdp\Indias GDP (Statewise).csv"
}

# === LOAD DATASETS ===
climate = pd.read_csv(files["climate"])
health = pd.read_csv(files["health"])
household_gdp = pd.read_excel(files["household_gdp"])
hospitals = pd.read_csv(files["hospitals"])
expenses = pd.read_csv(files["expenses"])
state_gdp = pd.read_csv(files["state_gdp"])

# === STANDARDIZE STATE COLUMN NAMES ===
climate.rename(columns={"statename": "state"}, inplace=True)
health.rename(columns={"state/ut": "state"}, inplace=True)
state_gdp.rename(columns={"state_ut": "state"}, inplace=True)
# others already have a column named 'state'

# === PICK ONLY USEFUL COLUMNS (optional) ===
climate = climate[["state", "totalpopulation", "totalpopulationurban"]].groupby("state").mean().reset_index()
health = health[["state", "number of households surveyed",
                 "female population age 6 years and above who ever attended school (%)",
                 "sex ratio of the total population (females per 1,000 males)"]].groupby("state").mean().reset_index()
state_gdp = state_gdp[["state", "nominal gdp(trillion inr)"]]
hospitals = hospitals.groupby("state").agg({"id": "count"}).reset_index().rename(columns={"id": "hospital_count"})

# For household_gdp (nationwide, no state info) — we'll use as reference later
household_gdp.rename(columns={"gdp": "india_total_gdp", "household_spending": "india_household_spending"}, inplace=True)

# Try to extract state avg expense from Monthly Expense file if possible
if "state" in expenses.columns:
    expenses = expenses[["state", "average_monthly_expense"]]
else:
    print("⚠️ 'state' column not found in expense file, check column names!")

# === MERGE ALL STATE-LEVEL DATA ===
merged = state_gdp.merge(expenses, on="state", how="outer")
merged = merged.merge(climate, on="state", how="outer")
merged = merged.merge(health, on="state", how="outer")
merged = merged.merge(hospitals, on="state", how="outer")

# === CLEANUP ===
merged = merged.drop_duplicates("state")
merged = merged.fillna(0)

# === SAVE MASTER FILE ===
out_path = r"E:\healthcare\datasets\merged_master.csv"
merged.to_csv(out_path, index=False)

print("✅ Merged dataset saved to:", out_path)
print("Shape:", merged.shape)
print(merged.head())
```

```python
import pandas as pd

# === FILE PATHS ===
files = {
    "climate": r"E:\healthcare\datasets\climate vulnerability\7160_KEYS.csv",
    "health": r"E:\healthcare\datasets\family health survey\national_family_health_survey.csv",
    "household_gdp": r"E:\healthcare\datasets\india gdp houshold\GDP and Household Consumption Expenditure Data for India (1990-2019).xlsx",
    "hospitals": r"E:\healthcare\datasets\india hospitals\Hospitals In India (Anonymized).csv",
    "expenses": r"E:\healthcare\datasets\monthly expense\Monthly_Expense_Data.csv",
    "state_gdp": r"E:\healthcare\datasets\state gdp\Indias GDP (Statewise).csv"
}

# === LOAD & CLEAN INDIVIDUAL FILES ===

# 1️⃣ CLIMATE
climate = pd.read_csv(files["climate"])
climate.rename(columns={
    "StateName": "state",
    "TotalPopulation": "total_population",
    "TotalPopulationUrban": "total_population_urban"
}, inplace=True)
climate = climate[["state", "total_population", "total_population_urban"]].groupby("state").mean().reset_index()

# 2️⃣ HEALTH
health = pd.read_csv(files["health"])
health.columns = health.columns.str.strip().str.lower()
health.rename(columns={"state/ut": "state"}, inplace=True)

# print columns so you can see them if needed
print("🩺 Health dataset columns:", list(health.columns))

# find best matches dynamically
possible_cols = [
    "state",
    "number of households surveyed",
    "female population age 6 years and above who ever attended school (%)",
    "sex ratio of the total population (females per 1,000 males)"
]

# attempt fuzzy matching of columns
selected_cols = []
for col in possible_cols:
    match = [c for c in health.columns if col.split()[0] in c]
    if match:
        selected_cols.append(match[0])

if "state" not in selected_cols and "state" in health.columns:
    selected_cols.append("state")

print("✅ Using health columns:", selected_cols)
health = health[selected_cols].groupby("state").mean().reset_index()

# 3️⃣ HOUSEHOLD GDP
household_gdp = pd.read_excel(files["household_gdp"])
household_gdp.rename(columns={
    "gdp": "india_total_gdp",
    "household_spending": "india_household_spending"
}, inplace=True)

# 4️⃣ HOSPITALS
hospitals = pd.read_csv(files["hospitals"])
hospitals = hospitals.groupby("state").agg({"id": "count"}).reset_index()
hospitals.rename(columns={"id": "hospital_count"}, inplace=True)

# 5️⃣ EXPENSES
expenses = pd.read_csv(files["expenses"])
# find state column
state_col = None
for col in expenses.columns:
    if "state" in col.lower():
        state_col = col
        break
if state_col:
    expenses.rename(columns={state_col: "state"}, inplace=True)
else:
    print("⚠️ Could not find state column in expenses file.")
# find expense column
expense_col = None
for col in expenses.columns:
    if "expense" in col.lower():
        expense_col = col
        break
if expense_col:
    expenses.rename(columns={expense_col: "average_monthly_expense"}, inplace=True)
expenses = expenses[["state", "average_monthly_expense"]]

# 6️⃣ STATE GDP
state_gdp = pd.read_csv(files["state_gdp"])
state_gdp.rename(columns={"state_ut": "state"}, inplace=True)
state_gdp = state_gdp[["state", "nominal gdp(trillion inr)"]]

# === MERGE ===
merged = state_gdp.merge(expenses, on="state", how="outer")
merged = merged.merge(climate, on="state", how="outer")
merged = merged.merge(health, on="state", how="outer")
merged = merged.merge(hospitals, on="state", how="outer")

# === CLEAN & SAVE ===
merged = merged.drop_duplicates("state").fillna(0)

out_path = r"E:\healthcare\datasets\merged_master.csv"
merged.to_csv(out_path, index=False)

print("\n✅ Merged dataset saved to:", out_path)
print("Shape:", merged.shape)
print(merged.head())
```

```python
import pandas as pd

# === FILE PATHS ===
files = {
    "climate": r"E:\healthcare\datasets\climate vulnerability\7160_KEYS.csv",
    "health": r"E:\healthcare\datasets\family health survey\national_family_health_survey.csv",
    "household_gdp": r"E:\healthcare\datasets\india gdp houshold\GDP and Household Consumption Expenditure Data for India (1990-2019).xlsx",
    "hospitals": r"E:\healthcare\datasets\india hospitals\Hospitals In India (Anonymized).csv",
    "expenses": r"E:\healthcare\datasets\monthly expense\Monthly_Expense_Data.csv",
    "state_gdp": r"E:\healthcare\datasets\state gdp\Indias GDP (Statewise).csv"
}

# === 1️⃣ CLIMATE ===
climate = pd.read_csv(files["climate"])
climate.rename(columns={
    "StateName": "state",
    "TotalPopulation": "total_population",
    "TotalPopulationUrban": "total_population_urban"
}, inplace=True)
climate = climate[["state", "total_population", "total_population_urban"]].groupby("state").mean().reset_index()

# === 2️⃣ HEALTH ===
health = pd.read_csv(files["health"])
health.columns = health.columns.str.strip().str.lower()
health.rename(columns={"state/ut": "state"}, inplace=True)
print("\n🩺 Health dataset columns:", list(health.columns))

possible_cols = [
    "state",
    "number of households surveyed",
    "female population age 6 years and above who ever attended school (%)",
    "sex ratio of the total population (females per 1,000 males)"
]

selected_cols = []
for col in possible_cols:
    match = [c for c in health.columns if col.split()[0] in c]
    if match:
        selected_cols.append(match[0])

if "state" not in selected_cols and "state" in health.columns:
    selected_cols.append("state")

print("✅ Using health columns:", selected_cols)
health = health[selected_cols].groupby("state").mean().reset_index()

# === 3️⃣ HOUSEHOLD GDP ===
household_gdp = pd.read_excel(files["household_gdp"])
household_gdp.rename(columns={
    "gdp": "india_total_gdp",
    "household_spending": "india_household_spending"
}, inplace=True)

# === 4️⃣ HOSPITALS ===
hospitals = pd.read_csv(files["hospitals"])
hospitals.columns = hospitals.columns.str.strip().str.lower()
print("\n🏥 Hospital dataset columns:", list(hospitals.columns))

# find correct column for state
state_col = None
for col in hospitals.columns:
    if "state" in col.lower():
        state_col = col
        break

if state_col:
    hospitals.rename(columns={state_col: "state"}, inplace=True)
    hospitals = hospitals.groupby("state").agg({"id": "count"}).reset_index()
    hospitals.rename(columns={"id": "hospital_count"}, inplace=True)
else:
    print("⚠️ No state column found in hospitals file. Skipping hospital aggregation.")
    hospitals = pd.DataFrame(columns=["state", "hospital_count"])

# === 5️⃣ EXPENSES ===
expenses = pd.read_csv(files["expenses"])
expenses.columns = expenses.columns.str.strip().str.lower()

state_col = None
for col in expenses.columns:
    if "state" in col:
        state_col = col
        break
if state_col:
    expenses.rename(columns={state_col: "state"}, inplace=True)

expense_col = None
for col in expenses.columns:
    if "expense" in col:
        expense_col = col
        break
if expense_col:
    expenses.rename(columns={expense_col: "average_monthly_expense"}, inplace=True)

if "state" in expenses.columns and "average_monthly_expense" in expenses.columns:
    expenses = expenses[["state", "average_monthly_expense"]]
else:
    print("⚠️ Expenses file missing state or expense column.")
    expenses = pd.DataFrame(columns=["state", "average_monthly_expense"])

# === 6️⃣ STATE GDP ===
state_gdp = pd.read_csv(files["state_gdp"])
state_gdp.columns = state_gdp.columns.str.strip().str.lower()
if "state_ut" in state_gdp.columns:
    state_gdp.rename(columns={"state_ut": "state"}, inplace=True)
state_gdp = state_gdp[["state", "nominal gdp(trillion inr)"]]

# === MERGE ===
merged = state_gdp.merge(expenses, on="state", how="outer")
merged = merged.merge(climate, on="state", how="outer")
merged = merged.merge(health, on="state", how="outer")
merged = merged.merge(hospitals, on="state", how="outer")

# === CLEAN & SAVE ===
merged = merged.drop_duplicates("state").fillna(0)
out_path = r"E:\healthcare\datasets\merged_master.csv"
merged.to_csv(out_path, index=False)

print("\n✅ Merged dataset saved to:", out_path)
print("Shape:", merged.shape)
print(merged.head())
```
